## Changing, Adding, and Removing Elements

### adding


- **append(...)** - add an element to the end of the list
- **insert(index, ...)** - add an element to a certain position; index out of range


```python
trans=['motorbike', 'car', 'bus']
trans.insert(6, 'plane')
print (trans)
```

#### Sort() method - làm thay đổi list
- sort theo alpha or nhỏ -> lớn (phải có mqh thứ tự - vd list vừa int vừa str => k sort đc)

```python
alist=[3,2,7,3,10,45]
alist.sort()
alist
```

- sắp xếp ngược lại - dùng tham số reverse

```python
alist=[3,2,7,3,10,45]
alist.sort(reverse=True)
alist
```

- sắp xếp theo độ dài của phần tử 

```python
alist=['3',[1,2],'2','afc']
alist.sort(key=lambda x:len(x)) #key- đặt biến ẩn danh 
alist
```

- sắp xếp theo tên 

- sắp xếp theo điểmm từ cao -> thấp

```python
scores = [['An', 8], ['Mai', 7], ['Anh', 10], ['Hà', 5]]
scores.sort(key=lambda x: x[0])
print(scores)
scores.sort(reverse=True, key=lambda x: x[1])
print(scores)
```

#### Sorted() function - thay đổi tạm thời


```python
print(sorted(scores))
```

```python
print(sorted(scores, key= lambda x: x[1], reverse=True))
```
